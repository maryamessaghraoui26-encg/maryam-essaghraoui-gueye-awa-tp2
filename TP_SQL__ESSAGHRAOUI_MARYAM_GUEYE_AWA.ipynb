```python
import pandas as pd

chemin_html = "/content/drive/MyDrive/Colab Notebooks/Morocco_ IMF's Holdings of Currency as of September 30, 2025 (3).html"

try:
    # Lire tous les tableaux du fichier HTML
    liste_tableaux = pd.read_html(chemin_html)

    # Recherche du tableau principal contenant la colonne 'Date'
    tableau_cible = None
    for idx, df_temp in enumerate(liste_tableaux):
        if any('date' in str(col).lower() for col in df_temp.columns):
            tableau_cible = df_temp
            print(f"Tableau de données trouvé à l'index {idx} !")
            break

    # Si trouvé, on l'affiche, sinon on affiche le tableau avec le plus de colonnes/lignes
    if tableau_cible is not None:
        display(tableau_cible.head())
    else:
        # Option de repli sur le tableau le plus grand qui contient souvent les données
        tableau_cible = max(liste_tableaux, key=lambda x: x.shape[0] * x.shape[1])
        display(tableau_cible.head())
except Exception as e:
    print("Erreur lors de l'extraction des données :", e)
```

```python
# On stocke le tableau cible extrait dans la variable principale 'df'
df = tableau_cible

# Affichage des dimensions et des premières lignes
print(f"Dimensions du DataFrame : {df.shape[0]} lignes et {df.shape[1]} colonnes.")
display(df.head())
```

```python
# 1. Afficher tous les relevés du Maroc (en utilisant pandas)
# Puisque les données chargées concernent déjà le Maroc, nous trions simplement par date de manière décroissante
if 'Date' in df.columns:
    # Conversion temporaire en datetime pour un tri chronologique correct si nécessaire, ou tri direct
    df_tri = df.sort_values(by='Date', ascending=False)
    display(df_tri)
else:
    display(df)
```

```python
# 2. Afficher les avoirs totaux au 30 septembre 2025 (Traduit en pandas)
# Nous filtrons sur la date 'September 30, 2025' et sélectionnons les colonnes d'intérêt
recherche_date = 'September 30, 2025'

if 'Date' in df.columns and 'Total Holdings' in df.columns:
    resultat = df[df['Date'] == recherche_date][['Date', 'Total Holdings']]
    display(resultat)
else:
    print("Les colonnes requises ne sont pas présentes dans le DataFrame.")
```

```python
# 3. Afficher les relevés de l'année 2024 (Traduit en pandas)
import pandas as pd

if 'Date' in df.columns:
    # Conversion de la colonne 'Date' au format datetime de pandas pour filtrer facilement par année
    df_temp_date = df.copy()
    df_temp_date['Date_parsed'] = pd.to_datetime(df_temp_date['Date'], errors='coerce')

    # Filtrer pour l'année 2024
    resultat_2024 = df_temp_date[df_temp_date['Date_parsed'].dt.year == 2024]

    # Trier par date de manière croissante (comme ORDER BY date_releve)
    resultat_2024 = resultat_2024.sort_values(by='Date_parsed', ascending=True)

    # Supprimer la colonne temporaire avant l'affichage
    resultat_2024 = resultat_2024.drop(columns=['Date_parsed'])

    display(resultat_2024)
else:
    print("La colonne 'Date' n'existe pas dans le DataFrame.")
```

```python
# 4. Trouver le relevé ayant le total d'avoirs le plus élevé (Traduit en pandas)
if 'Date' in df.columns and 'Total Holdings' in df.columns:
    # Trouver l'index de la ligne ayant la valeur maximale dans 'Total Holdings'
    index_max = df['Total Holdings'].idxmax()

    # Extraire et afficher la date et le total d'avoirs correspondant
    resultat_max = df.loc[[index_max], ['Date', 'Total Holdings']]
    display(resultat_max)
else:
    print("Les colonnes requises ne sont pas présentes dans le DataFrame.")
```

```python
# 5. Calculer la moyenne des avoirs totaux (Traduit en pandas)
if 'Total Holdings' in df.columns:
    # Calcul de la moyenne de la colonne 'Total Holdings'
    moyenne_avoirs = df['Total Holdings'].mean()
    print(f"Moyenne des avoirs totaux : {moyenne_avoirs:,.2f}")
else:
    print("La colonne 'Total Holdings' n'est pas présente dans le DataFrame.")
```
